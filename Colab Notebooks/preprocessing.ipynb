```python
import os
import shutil
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Base directory
base_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/Dataset"
output_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset"

# Define input and output paths
input_images_dir = os.path.join(base_dir, "Images")
input_labels_dir = os.path.join(base_dir, "Labels")
output_images_train = os.path.join(output_dir, "images/train")
output_images_val = os.path.join(output_dir, "images/val")
output_labels_train = os.path.join(output_dir, "labels/train")
output_labels_val = os.path.join(output_dir, "labels/val")

# Create output directories
os.makedirs(output_images_train, exist_ok=True)
os.makedirs(output_images_val, exist_ok=True)
os.makedirs(output_labels_train, exist_ok=True)
os.makedirs(output_labels_val, exist_ok=True)

def move_files(source_dir, dest_dir, file_extension):
    """
    Moves all files with a specific extension from subdirectories into a destination directory.
    """
    for root, _, files in os.walk(source_dir):
        for file in files:
            if file.endswith(file_extension):
                shutil.copy2(os.path.join(root, file), os.path.join(dest_dir, file))

# Move train and val images and labels
move_files(os.path.join(input_images_dir, "train"), output_images_train, ".jpg")
move_files(os.path.join(input_images_dir, "val"), output_images_val, ".jpg")
move_files(os.path.join(input_labels_dir, "train"), output_labels_train, ".txt")
move_files(os.path.join(input_labels_dir, "val"), output_labels_val, ".txt")

print("Dataset successfully organized for YOLO format!")
```

```python
import os
import random
import cv2
import numpy as np
from PIL import Image, ImageEnhance
from pathlib import Path
from tqdm import tqdm

# Augmentation Functions
def add_salt_and_pepper_noise(image, salt_prob=0.01, pepper_prob=0.01):
    """Add salt and pepper noise to the image."""
    noisy_image = image.copy()
    total_pixels = noisy_image.size
    num_salt = int(total_pixels * salt_prob)
    num_pepper = int(total_pixels * pepper_prob)

    # Salt noise (white pixels)
    salt_coords = [random.randint(0, i - 1) for i in noisy_image.shape]
    for coord in salt_coords:
        noisy_image[coord] = [255, 255, 255]

    # Pepper noise (black pixels)
    pepper_coords = [random.randint(0, i - 1) for i in noisy_image.shape]
    for coord in pepper_coords:
        noisy_image[coord] = [0, 0, 0]

    return noisy_image

def apply_gaussian_blur(image, kernel_size=(5, 5)):
    """Apply Gaussian Blur to smooth the image."""
    return cv2.GaussianBlur(image, kernel_size, 0)

def adjust_brightness_contrast(image, brightness_factor=1.2, contrast_factor=1.2):
    """Adjust brightness and contrast randomly."""
    pil_image = Image.fromarray(image)  # Create PIL image
    enhancer = ImageEnhance.Brightness(pil_image)
    pil_image = enhancer.enhance(brightness_factor)
    enhancer = ImageEnhance.Contrast(pil_image)
    pil_image = enhancer.enhance(contrast_factor)
    return np.array(pil_image)

def random_sharpness(image, sharpness_factor=2.0):
    """Apply random sharpness adjustment."""
    pil_image = Image.fromarray(image)
    enhancer = ImageEnhance.Sharpness(pil_image)
    pil_image = enhancer.enhance(sharpness_factor)
    return np.array(pil_image)

def color_jitter(image, hue_shift=10, saturation_shift=30, value_shift=30):
    """Apply random color jitter (hue, saturation, value)."""
    hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv_image)

    # Adjust hue, saturation, and value
    h = np.uint8((h + random.randint(-hue_shift, hue_shift)) % 180)
    s = np.clip(s + random.randint(-saturation_shift, saturation_shift), 0, 255).astype(np.uint8)
    v = np.clip(v + random.randint(-value_shift, value_shift), 0, 255).astype(np.uint8)

    # Merge and convert back to BGR
    hsv_jittered = cv2.merge([h, s, v])
    return cv2.cvtColor(hsv_jittered, cv2.COLOR_HSV2BGR)

def apply_augmentations(image):
    """Apply all augmentations to an image."""
    image = add_salt_and_pepper_noise(image)
    image = apply_gaussian_blur(image)
    image = adjust_brightness_contrast(image)
    image = random_sharpness(image)
    image = color_jitter(image)
    return image

def process_labels(label_file, augmented_image_filename, label_folder):
    """Process label files to ensure they match the augmented images."""
    with open(label_file, 'r') as f:
        lines = f.readlines()

    augmented_label_file = os.path.join(label_folder, augmented_image_filename.replace('.jpg', '.txt'))

    # Ensure the directory exists
    os.makedirs(os.path.dirname(augmented_label_file), exist_ok=True)

    with open(augmented_label_file, 'w') as f:
        for line in lines:
            # Each line has format: <class> <x_center> <y_center> <width> <height>
            f.write(line)

def process_images_and_labels(images_dir, labels_dir, augmented_images_dir, augmented_labels_dir):
    """Process all images and labels and apply augmentations."""
    image_files = [f for f in os.listdir(images_dir) if f.endswith('.jpg')]
    if not image_files:
        print(f"No image files found in {images_dir}")
        return
    for image_file in tqdm(image_files, desc="Processing Images"):
        # Read image and label
        image_path = os.path.join(images_dir, image_file)
        if not os.path.exists(image_path):
            print(f"Image not found: {image_path}")
            continue

        image = cv2.imread(image_path)
        label_file = os.path.join(labels_dir, image_file.replace('.jpg', '.txt'))
        if not os.path.exists(label_file):
            print(f"Label not found for {image_file}")
            continue

        # Apply augmentations and save the augmented images and labels
        augmented_image = apply_augmentations(image)

        # Save augmented image
        augmented_image_filename = image_file.replace('.jpg', '_augmented.jpg')
        cv2.imwrite(os.path.join(augmented_images_dir, augmented_image_filename), augmented_image)

        # Process the label file for augmented image
        process_labels(label_file, augmented_image_filename, augmented_labels_dir)

def create_yolo_dataset(images_dir, labels_dir, augmented_images_dir, augmented_labels_dir):
    """Create YOLO dataset with augmentations."""
    Path(augmented_images_dir).mkdir(parents=True, exist_ok=True)
    Path(augmented_labels_dir).mkdir(parents=True, exist_ok=True)

    # Process training and validation datasets
    print(f"Processing training dataset from {images_dir}/train")
    process_images_and_labels(
        os.path.join(images_dir, "train"),
        os.path.join(labels_dir, "train"),
        os.path.join(augmented_images_dir, "train"),
        os.path.join(augmented_labels_dir, "train")
    )

    print(f"Processing validation dataset from {images_dir}/val")
    process_images_and_labels(
        os.path.join(images_dir, "val"),
        os.path.join(labels_dir, "val"),
        os.path.join(augmented_images_dir, "val"),
        os.path.join(augmented_labels_dir, "val")
    )

# Example usage:
images_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset/images"
labels_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset/labels"
augmented_images_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset/augmented_images"
augmented_labels_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset/augmented_labels"

create_yolo_dataset(images_dir, labels_dir, augmented_images_dir, augmented_labels_dir)
```

```python
import os
import cv2
import numpy as np
from albumentations import Compose, RandomBrightnessContrast, GaussNoise, Blur, RandomGamma, CLAHE
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Base directory
base_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset"
images_train_dir = os.path.join(base_dir, "images/train")
labels_train_dir = os.path.join(base_dir, "labels/train")

# Define output directories for augmented data
augmented_images_dir = os.path.join(base_dir, "augmented_images")
augmented_labels_dir = os.path.join(base_dir, "augmented_labels")

# Create output directories
os.makedirs(augmented_images_dir, exist_ok=True)
os.makedirs(augmented_labels_dir, exist_ok=True)

# Define augmentation pipeline (no geometric transformations)
transform = Compose([
    RandomBrightnessContrast(p=0.2),
    GaussNoise(var_limit=(10, 50), p=0.2),
    Blur(blur_limit=3, p=0.2),
    RandomGamma(gamma_limit=(80, 120), p=0.2),
    CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=0.2)
])

def augment_image_and_label(image_path, label_path, output_image_path, output_label_path):
    try:
        # Read the image
        image = cv2.imread(image_path)
        if image is None:
            print(f"Warning: Failed to read {image_path}. Skipping.")
            return

        # Read the label
        if not os.path.exists(label_path):
            print(f"Warning: Missing label for {image_path}. Skipping.")
            return

        with open(label_path, 'r') as file:
            label_data = file.readlines()

        # Apply augmentation
        augmented = transform(image=image)
        augmented_image = augmented['image']

        # Save augmented image
        cv2.imwrite(output_image_path, augmented_image)

        # Save the label (labels remain the same)
        with open(output_label_path, 'w') as file:
            file.writelines(label_data)

    except Exception as e:
        print(f"Error processing {image_path}: {e}")

def generate_augmented_images_and_labels():
    processed_count = 0
    for image_file in os.listdir(images_train_dir):
        if image_file.endswith(".jpg"):
            image_path = os.path.join(images_train_dir, image_file)
            label_file = image_file.replace(".jpg", ".txt")
            label_path = os.path.join(labels_train_dir, label_file)

            # Define output paths
            output_image_path = os.path.join(augmented_images_dir, f"aug_{image_file}")
            output_label_path = os.path.join(augmented_labels_dir, f"aug_{label_file}")

            # Augment and save the image/label pair
            augment_image_and_label(image_path, label_path, output_image_path, output_label_path)
            processed_count += 1

    print(f"Augmentation completed. Total processed files: {processed_count}")

# Perform augmentation on images and labels
generate_augmented_images_and_labels()
```

```python
import os
import cv2
from albumentations import Compose, RandomBrightnessContrast, GaussNoise, Blur, RandomGamma, CLAHE
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Base directories
base_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset"
images_train_dir = os.path.join(base_dir, "images/train")
labels_train_dir = os.path.join(base_dir, "labels/train")
images_val_dir = os.path.join(base_dir, "images/val")
labels_val_dir = os.path.join(base_dir, "labels/val")

# Output directories for augmented data
aug_images_train_dir = os.path.join(base_dir, "augmented_images/train")
aug_labels_train_dir = os.path.join(base_dir, "augmented_labels/train")
aug_images_val_dir = os.path.join(base_dir, "augmented_images/val")
aug_labels_val_dir = os.path.join(base_dir, "augmented_labels/val")

# Create output directories
os.makedirs(aug_images_train_dir, exist_ok=True)
os.makedirs(aug_labels_train_dir, exist_ok=True)
os.makedirs(aug_images_val_dir, exist_ok=True)
os.makedirs(aug_labels_val_dir, exist_ok=True)

# Augmentation pipeline
transform = Compose([
    RandomBrightnessContrast(p=0.2),  # Adjust brightness/contrast
    GaussNoise(var_limit=(10, 50), p=0.2),  # Add Gaussian noise
    Blur(blur_limit=3, p=0.2),  # Apply blur
    RandomGamma(gamma_limit=(80, 120), p=0.2),  # Randomly adjust gamma
    CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=0.2)  # Apply CLAHE for contrast enhancement
])

def augment_image_and_label(image_path, label_path, output_image_dir, output_label_dir, num_augmentations=4):
    # Read the image
    image = cv2.imread(image_path)

    # Read the label
    with open(label_path, 'r') as file:
        label_data = file.readlines()

    # Generate augmented images
    base_filename = os.path.basename(image_path).split(".")[0]
    for i in range(num_augmentations):
        # Apply augmentation
        augmented = transform(image=image)
        augmented_image = augmented['image']

        # Define output file paths
        augmented_image_path = os.path.join(output_image_dir, f"{base_filename}_aug{i+1}.jpg")
        augmented_label_path = os.path.join(output_label_dir, f"{base_filename}_aug{i+1}.txt")

        # Save augmented image
        cv2.imwrite(augmented_image_path, augmented_image)

        # Save the label (unchanged)
        with open(augmented_label_path, 'w') as file:
            for line in label_data:
                file.write(line)

def process_folder(images_dir, labels_dir, output_images_dir, output_labels_dir, num_augmentations=4):
    for image_file in os.listdir(images_dir):
        if image_file.endswith(".jpg"):
            image_path = os.path.join(images_dir, image_file)
            label_file = image_file.replace(".jpg", ".txt")
            label_path = os.path.join(labels_dir, label_file)

            # Ensure the corresponding label exists
            if os.path.exists(label_path):
                augment_image_and_label(image_path, label_path, output_images_dir, output_labels_dir, num_augmentations)

# Perform augmentation for training and validation datasets
print("Processing training dataset...")
process_folder(images_train_dir, labels_train_dir, aug_images_train_dir, aug_labels_train_dir, num_augmentations=4)

print("Processing validation dataset...")
process_folder(images_val_dir, labels_val_dir, aug_images_val_dir, aug_labels_val_dir, num_augmentations=4)

print("Augmentation completed and saved.")
```

```python
import os
import shutil

# Define directories
base_dir = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset"
aug_images_train_dir = os.path.join(base_dir, "augmented_images/train")
aug_labels_train_dir = os.path.join(base_dir, "augmented_labels/train")
aug_images_val_dir = os.path.join(base_dir, "augmented_images/val")
aug_labels_val_dir = os.path.join(base_dir, "augmented_labels/val")

images_train_dir = os.path.join(base_dir, "images/train")
labels_train_dir = os.path.join(base_dir, "labels/train")
images_val_dir = os.path.join(base_dir, "images/val")
labels_val_dir = os.path.join(base_dir, "labels/val")

def merge_folders(source_dir, target_dir):
    """Move all files from source_dir to target_dir."""
    if not os.path.exists(source_dir):
        print(f"Source directory not found: {source_dir}")
        return
    os.makedirs(target_dir, exist_ok=True)
    for file_name in os.listdir(source_dir):
        source_path = os.path.join(source_dir, file_name)
        target_path = os.path.join(target_dir, file_name)
        if os.path.isfile(source_path):
            shutil.move(source_path, target_path)
    print(f"Merged files from {source_dir} to {target_dir}")

# Merge augmented train images and labels
merge_folders(aug_images_train_dir, images_train_dir)
merge_folders(aug_labels_train_dir, labels_train_dir)

# Merge augmented val images and labels
merge_folders(aug_images_val_dir, images_val_dir)
merge_folders(aug_labels_val_dir, labels_val_dir)

print("Merging completed successfully.")
```
