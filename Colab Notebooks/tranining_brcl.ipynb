```python
import os
from ultralytics import YOLO

# Set the path for your YOLO dataset (adjust the paths based on your setup)
dataset_path = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset"
data_yaml_path = os.path.join(dataset_path, "data.yaml")

# Path to the pre-trained YOLO model (you can choose a model from yolov8n, yolov8s, yolov8m, etc.)
model_path = "yolov8n.pt"  # You can use yolov8n, yolov8s, yolov8m, yolov8l, etc.

# Set the training parameters
epochs = 50  # Number of training epochs
batch_size = 16  # Batch size for training
img_size = 640  # Image size (default is usually 640x640)
device = 0  # Use GPU (0 for the first GPU, 'cpu' for CPU training)

# Ensure the paths and data.yaml are correct
assert os.path.exists(data_yaml_path), f"Data YAML file not found at {data_yaml_path}"

# Initialize YOLO model with the pre-trained weights
model = YOLO(model_path)

# Define the output folder for saving the trained model
output_model_folder = "/content/drive/MyDrive/Final_Broccolli_Detection_ML/YOLO_Dataset/trained_model1.0"

# Ensure the output folder exists
os.makedirs(output_model_folder, exist_ok=True)

# Train the model
model.train(
    data=data_yaml_path,  # Path to the data.yaml file
    epochs=epochs,  # Number of epochs
    batch=batch_size,  # Batch size
    imgsz=img_size,  # Image size (height x width)
    device=device,  # Set device to '0' for GPU or 'cpu' for CPU
    project=output_model_folder,  # Save the results in this folder
    name='broccoli_model',  # Name of the project or folder to store results
    exist_ok=True  # If the folder exists, overwrite it
)

# Output the path to the saved model
saved_model_path = os.path.join(output_model_folder, 'broccoli_model', 'weights', 'best.pt')
print(f"Training complete. Model saved at: {saved_model_path}")
```
